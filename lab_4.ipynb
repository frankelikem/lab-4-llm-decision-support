````python
import os
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.environ["GROQ_API_KEY"]

print("API key loaded successfully.")
````

````python
from openai import OpenAI

client = OpenAI(
    api_key=API_KEY,
    base_url="https://api.groq.com/openai/v1"
)

MODEL = "llama-3.3-70b-versatile"

print("Client ready.")
````

````python
def ask_llm(
    user_prompt,
    system_prompt="You are a helpful assistant.",
    temperature=0.7,
    max_tokens=500
):
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt},
        ],
        temperature=temperature,
        max_tokens=max_tokens,
    )

    return response
````

````python
response = ask_llm(
    "What are three benefits of saving money?"
)

print(response.choices[0].message.content)
print("\nToken usage:")
print(response.usage)
````

````python
EXTRACT_SYSTEM_PROMPT = """
You are a data extraction assistant for a microfinance loan officer.

Extract information from the loan application and return ONLY a valid JSON object.

The JSON must contain EXACTLY these keys:

{
  "applicant_name": "string",
  "amount_ghs": "number",
  "purpose": "string",
  "monthly_profit_ghs": "number or null",
  "has_collateral_or_guarantor": "boolean",
  "repayment_months": "number or null"
}

Rules:
- Extract only information explicitly stated in the application.
- Do not guess or infer missing information.
- If a field is not stated, use null.
- amount_ghs must be a number, not a string.
- monthly_profit_ghs must be a number or null.
- repayment_months must be a number or null.
- has_collateral_or_guarantor is true if the applicant explicitly mentions
  collateral or a guarantor; otherwise false.
- Return ONLY JSON. Do not include explanations or markdown.
"""
````

````python
EXAMPLE_LETTER = """
My name is Ama. I run a small bakery and have operated for five years.
I am requesting GHS 6,000 to buy a new oven. My monthly profit is GHS 1,200.
My brother will guarantee the loan. I plan to repay it over 10 months.
"""

EXAMPLE_JSON = """
{
  "applicant_name": "Ama",
  "amount_ghs": 6000,
  "purpose": "buy a new oven",
  "monthly_profit_ghs": 1200,
  "has_collateral_or_guarantor": true,
  "repayment_months": 10
}
"""
````

````python
EXTRACT_PROMPT = """
Extract the required information from the loan application below.

Example:

Loan application:
{example_letter}

Correct JSON:
{example_json}

Now extract information from this application:

Loan application:
{letter_text}

Return ONLY the JSON object.
"""
````

````python
import json

def extract_fields(letter_text, temperature=0.0):
    prompt = EXTRACT_PROMPT.format(
        example_letter=EXAMPLE_LETTER,
        example_json=EXAMPLE_JSON,
        letter_text=letter_text
    )

    try:
        response = ask_llm(
            prompt,
            system_prompt=EXTRACT_SYSTEM_PROMPT,
            temperature=temperature,
            max_tokens=300
        )

        result = response.choices[0].message.content.strip()

        # Remove markdown JSON fences if present
        if result.startswith("```json"):
            result = result[7:]

        if result.startswith("```"):
            result = result[3:]

        if result.endswith("```"):
            result = result[:-3]

        result = result.strip()

        return json.loads(result)

    except json.JSONDecodeError:
        print("WARNING: Model returned invalid JSON.")
        print("Raw response:", result)
        return None

    except Exception as e:
        print("WARNING:", e)
        return None
````

````python
BRIEF_SYSTEM_PROMPT = """
You are an assistant supporting a microfinance loan officer.

Your job is to analyze a loan application and provide a neutral
decision-support brief.

The final loan decision must always be made by a human loan officer.

Use ONLY information contained in the loan application and extracted data.
Do not invent facts or make assumptions.

Your response must contain exactly these sections:

1. Strengths
- List positive factors supported by the application.

2. Risks / Red flags
- List relevant financial, business, repayment, or other risks supported
  by the application.
- Do not exaggerate or invent risks.

3. Missing information
- List important information that is not provided and should be requested
  by the loan officer.

4. Suggested next step
- Recommend an appropriate action such as requesting documents,
  inviting the applicant for an interview, or flagging the application
  for senior review.
- NEVER say "approve" or "reject".
"""
````

````python
BRIEF_PROMPT = """
Prepare a decision-support brief for the following loan application.

ORIGINAL APPLICATION:
{letter_text}

EXTRACTED INFORMATION:
{extracted_json}

Remember: the final loan decision must be made by a human.
Do not approve or reject the application.
"""
````

````python
import json

def generate_brief(letter_text, extracted_data):
    prompt = BRIEF_PROMPT.format(
        letter_text=letter_text,
        extracted_json=json.dumps(extracted_data, indent=2)
    )

    response = ask_llm(
        prompt,
        system_prompt=BRIEF_SYSTEM_PROMPT,
        temperature=0.0,
        max_tokens=500
    )

    return response.choices[0].message.content
````

````python
briefs = {}

for letter_id in LETTERS:
    extracted_row = extracted_df[
        extracted_df["letter_id"] == letter_id
    ].iloc[0].to_dict()

    briefs[letter_id] = generate_brief(
        LETTERS[letter_id],
        extracted_row
    )

print("Generated briefs for all six applications.")
````

````python
# Fields we need to evaluate
fields = [
    "applicant_name",
    "amount_ghs",
    "purpose",
    "monthly_profit_ghs",
    "has_collateral_or_guarantor",
    "repayment_months"
]

gold_ids = ["L001", "L003", "L006"]

accuracy_results = []

for field in fields:
    row = {"field": field}
    correct = 0

    for letter_id in gold_ids:
        predicted = extracted_df.loc[
            extracted_df["letter_id"] == letter_id, field
        ].iloc[0]

        expected = GOLD[letter_id][field]

        # Names are case-insensitive
        if field == "applicant_name":
            is_correct = (
                str(predicted).strip().lower()
                == str(expected).strip().lower()
            )

        # Purposes can have slightly different wording,
        # so compare after normalizing the text.
        elif field == "purpose":
            predicted_text = str(predicted).lower().strip()
            expected_text = str(expected).lower().strip()

            is_correct = (
                predicted_text == expected_text
                or expected_text in predicted_text
                or predicted_text in expected_text
            )

        else:
            is_correct = predicted == expected

        row[letter_id] = is_correct

        if is_correct:
            correct += 1

    row["accuracy"] = correct / len(gold_ids)
    accuracy_results.append(row)

accuracy_df = pd.DataFrame(accuracy_results)

display(accuracy_df)
````

````python
import json

def run_reliability_test(letter_id, temperature, runs=5):
    results = []
    valid_results = []

    for i in range(runs):
        result = extract_fields(
            LETTERS[letter_id],
            temperature=temperature
        )

        results.append(result)

        if result is not None:
            valid_results.append(
                json.dumps(result, sort_keys=True)
            )

    unique_results = set(valid_results)

    return {
        "temperature": temperature,
        "total_runs": runs,
        "valid_json": len(valid_results),
        "unique_valid_outputs": len(unique_results),
        "identical_outputs": (
            len(valid_results) > 0 and len(unique_results) == 1
        ),
        "results": results
    }
````

````python
temp_0_results = run_reliability_test(
    "L004",
    temperature=0.0
)

temp_1_results = run_reliability_test(
    "L004",
    temperature=1.0
)

print("===== TEMPERATURE 0.0 =====")
print("Valid JSON:", temp_0_results["valid_json"])
print("Unique valid outputs:", temp_0_results["unique_valid_outputs"])
print("All identical:", temp_0_results["identical_outputs"])

print("\n===== TEMPERATURE 1.0 =====")
print("Valid JSON:", temp_1_results["valid_json"])
print("Unique valid outputs:", temp_1_results["unique_valid_outputs"])
print("All identical:", temp_1_results["identical_outputs"])
````

Could automated lending be unfair?
If the bank fully automated the decisions, applicants who write poorly in English could be unfairly rejected even if they operate successful businesses. The model may mistake poor writing ability for poor business ability or financial reliability. This could disadvantage applicants with limited English proficiency or different communication styles. Human review is therefore important so that the loan decision considers the applicant's actual business situation rather than only the quality of their writing.

Sending the letters to a third-party API?
Loan applications contain personal and potentially sensitive financial information. Sending this information to a third-party API in another country could create privacy, security, and data-protection risks. Before deploying the system at a real Ghanaian microfinance institution, I would check how the API provider stores and processes data, whether the data can be used for model training, where the data is stored, what security measures are provided, and whether the institution has the necessary legal and organizational permissions to send the information to that provider.

Two production safeguards?
1. Human review: The LLM should never make the final loan decision. A qualified loan officer should review the original application, the extracted information, and the LLM's recommendations before making a decision.

2. Monitoring and logging: The system should record its outputs and monitor for errors, hallucinations, inconsistent results, and potential bias. Regular audits could identify problems before they cause widespread harm.

1. Prompting as engineering

Iterating on prompts is similar to tuning model hyperparameters because both involve changing settings or instructions, testing the results, and evaluating whether performance improves. The difference is that prompting changes how an existing model is instructed, while hyperparameter tuning in a trained model changes aspects of how the model learns or behaves during training.

2. Trust

I would not trust this system to run completely unattended. Although the evaluation can show good accuracy and consistency, the hallucination tests demonstrate that an LLM can still produce unsupported information. The extraction accuracy and hallucination results are therefore important reasons to keep a human involved in the final loan decision.

4. API vs. training your own model

Calling a foundation model through an API is more practical for this task because training a large language model from scratch would require much more data, computing power, time, and expertise. An API provides access to an already-trained model and allows the system to be developed quickly. Training a model could be more appropriate when an organization has a large specialized dataset, strict requirements for customization, or a need to control the model and its infrastructure directly.
